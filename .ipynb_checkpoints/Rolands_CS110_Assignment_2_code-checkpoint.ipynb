# Problem Set - Algorithmic Strategies

## Python Implementation

### Brute Force

```python
def bruteforce_search_extension(nums, target):
    ''' 
    In an ordered list of houses, finds the street numbers of the first and last house with the target postal code
    and counts how many times the target postal code appears in the list using a brute-force implementation.

    Parameters
    ----------
    nums : list of int
        A sorted list of integers representing postal codes of each house.
    target : int
        Target postal code to find in the list.

    Returns
    ----------
    list or None
        A list consisting of the street number of the first and last house with the target postal code and 
        the number of times the target postal code appers in the list. If the target is not found, returns None.
    '''

    # Notes street number of first and last houses with target.
    first_house = None
    last_house = None 

    # Iterates through list of houses and note street numbers of first and last houses with target if found.
    for street_number, postal_code in enumerate(nums): 
        if postal_code == target:              
            if first_house == None: # Check whether street number of first house was noted yet; if not, record street number.
                first_house = street_number
            last_house = street_number # Update street number of last house with target.

    # Calculates total number of houses with target if target is found; return answers.
    if first_house == None:
        return None
    else:
        total_houses = last_house - first_house + 1
        return [first_house, last_house, total_houses]
```

### Iterative Binary Search

```python
def iterative_binary_search_extension(nums, target):
    ''' 
    In an ordered list of houses, finds the street numbers of the first and last house with the target postal code
    and counts how many times the target postal code appears in the list using an iterative binary search implementation.

    Parameters
    ----------
    nums : list of int
        A sorted list of integers representing postal codes of each house.
    target : int
        Target postal code to find in the list.

    Returns
    ----------
    list or None
        A list consisting of the street number of the first and last house with the target postal code and 
        the number of times the target postal code appers in the list. If the target is not found, returns None.
    '''
    # Initializes bounds for search space.
    left_bound = 0
    right_bound = len(nums) - 1

    # Notes street number of first and last houses with target.
    first_house = None
    last_house = None

    # Finds street number of first house with target code.
    while left_bound <= right_bound:
        middle_house = (left_bound + right_bound) // 2
        if nums[middle_house] == target:
            first_house = middle_house
            right_bound = middle_house - 1
        elif nums[middle_house] < target:
            left_bound = middle_house + 1
        else:
            right_bound = middle_house - 1
    
    left_bound = 0
    right_bound = len(nums) - 1

    # Finds street number of last house with target code.
    while left_bound <= right_bound:
        middle_house = (left_bound + right_bound) // 2
        if nums[middle_house] == target:
            last_house = middle_house
            left_bound = middle_house + 1
        elif nums[middle_house] < target:
            left_bound = middle_house + 1
        else:
            right_bound = middle_house - 1

    # Returns None if target is not in list of houses.
    if first_house == None:
        return None

    # Calculates number of houses with target code in list.
    total_houses = last_house - first_house + 1

    return [first_house, last_house, total_houses]
```

### Recursive Binary Search

```python
def recursive_binary_search_extension(nums, target):
    ''' 
    In an ordered list of houses, finds the street numbers of the first and last house with the target postal code
    and counts how many times the target postal code appears in the list using a recursive binary search implementation.

    Parameters
    ----------
    nums : list of int
        A sorted list of integers representing postal codes of each house.
    target : int
        Target postal code to find in the list.

    Returns
    ----------
    list or None
        A list consisting of the street number of the first and last house with the target postal code and 
        the number of times the target postal code appers in the list. If the target is not found, returns None.
    '''

    # Recursive function to find first house with target postal code.
    def find_first_house(nums, target, left_bound, right_bound):
        '''
        Recursively finds the street number of the first house with the target in a sorted list.

        Parameters
        ----------
        nums : list of int
            A sorted list of integers representing postal codes of each house.
        target : int
            The postal code to find in the list.
        left_bound : int
            The left boundary of the search space.
        right_bound : int
            The right boundary of the search space.

        Returns
        ----------
        int or None
            The street number of the first house with the target if found, otherwise None.
        '''

        # Base case: Returns None if gone beyond bounds.
        if left_bound > right_bound:
            return None

        # Calculates middle house street number.
        middle_house = (left_bound + right_bound) // 2

        if nums[middle_house] == target:
            # Searches left to find street number of first house with target.
            first_house = find_first_house(nums, target, left_bound, middle_house - 1)

            # If another house with target is found on the left, returns the street number of that house. 
            # Otherwise, returns middle house street number.
            if first_house != None:
                return first_house
            else:
                return middle_house
                
        elif nums[middle_house] > target:
            return find_first_house(nums, target, left_bound, middle_house - 1)
        else:
            return find_first_house(nums, target, middle_house + 1, right_bound)

    # Recursive function to find last house with target postal code.
    def find_last_house(nums, target, left_bound, right_bound):
        '''
        Recursively finds the street number of the last house with the target in a sorted list.

        Parameters
        ----------
        nums : list of int
            A sorted list of integers representing postal codes of each house.
        target : int
            The postal code to find in the list.
        left_bound : int
            The left boundary of the search space.
        right_bound : int
            The right boundary of the search space.

        Returns
        ----------
        int or None
            The street number of the last house with the target if found, otherwise None.
        '''

        # Base case: Returns None if gone beyond bounds.
        if left_bound > right_bound:
            return None

        # Calculates middle house street number.
        middle_house = (left_bound + right_bound) // 2

        if nums[middle_house] == target:
            # Searches right to find street number of first house with target.
            last_house = find_last_house(nums, target, middle_house + 1, right_bound)

            # If another house with target is found on the right, return the street number of that house. 
            # Otherwise, return middle house street number.
            if last_house != None:
                return last_house
            else:
                return middle_house 
        elif nums[middle_house] > target:
            return find_last_house(nums, target, left_bound, middle_house - 1)
        else:
            return find_last_house(nums, target, middle_house + 1, right_bound)

    # Finds street number of first house with target code.
    first_house = find_first_house(nums, target, 0, len(nums) - 1)

    # Finds street number of first house with target code and calculates total number of houses with street number.
    last_house = find_last_house(nums, target, 0, len(nums) - 1)

    # Returns None if no house with target code is found.
    if first_house == None or last_house == None:
        return None
        
    total_houses = last_house - first_house + 1

    # Answer.
    return [first_house, last_house, total_houses]
```

### Iterative Ternary Search

```python
def iterative_ternary_search_extension(nums, target):
    ''' 
    In an ordered list of houses, finds the street numbers of the first and last house with the target postal code
    and counts how many times the target postal code appears in the list using an iterative ternary search implementation.

    Parameters
    ----------
    nums : list of int
        A sorted list of integers representing postal codes of each house.
    target : int
        Target postal code to find in the list.

    Returns
    ----------
    list or None
        A list consisting of the street number of the first and last house with the target postal code and 
        the number of times the target postal code appers in the list. If the target is not found, returns None.
    '''

    # Initializes bounds for search space.
    left_bound = 0
    right_bound = len(nums) - 1

    # Notes street number of first and last houses with target.
    first_house = None
    last_house = None

    # Finds street number of first house with target code.
    while left_bound <= right_bound:
        # Calculates two middle house street numbers.
        middle_house_1 = left_bound + (right_bound - left_bound) // 3
        middle_house_2 = right_bound - (right_bound - left_bound) // 3

        if nums[middle_house_1] == target:
            first_house = middle_house_1
            right_bound = middle_house_1 - 1
        elif nums[middle_house_2] == target:
            first_house = middle_house_2
            right_bound = middle_house_2 - 1
        elif target < nums[middle_house_1]:
            right_bound = middle_house_1 - 1
        elif target > nums[middle_house_2]:
            left_bound = middle_house_2 + 1
        else:
            left_bound = middle_house_1 + 1
            right_bound = middle_house_2 - 1

    # Re-initialize search space.
    left_bound = 0
    right_bound = len(nums) - 1

    # Finds street number of last house with target code.
    while left_bound <= right_bound:
        # Calculates two middle house street numbers.
        middle_house_1 = left_bound + (right_bound - left_bound) // 3
        middle_house_2 = right_bound - (right_bound - left_bound) // 3

        if nums[middle_house_2] == target:
            last_house = middle_house_2
            left_bound = middle_house_2 + 1 
        elif nums[middle_house_1] == target:
            last_house = middle_house_1
            left_bound = middle_house_1 + 1  
        elif target < nums[middle_house_1]:
            right_bound = middle_house_1 - 1
        elif target > nums[middle_house_2]:
            left_bound = middle_house_2 + 1
        else:
            left_bound = middle_house_1 + 1
            right_bound = middle_house_2 - 1

    # Returns None if target is not in list of houses.
    if first_house == None:
        return None

    # Calculates number of houses with target in list.
    total_houses = last_house - first_house + 1

    # Answer.
    return [first_house, last_house, total_houses]
```

### Recursive Ternary Search

```python
def recursive_ternary_search_extension(nums, target):
    ''' 
    In an ordered list of houses, finds the street numbers of the first and last house with the target postal code
    and counts how many times the target postal code appears in the list using a recursive ternary search implementation.

    Parameters
    ----------
    nums : list of int
        A sorted list of integers representing postal codes of each house.
    target : int
        Target postal code to find in the list.

    Returns
    ----------
    list or None
        A list consisting of the street number of the first and last house with the target postal code and 
        the number of times the target postal code appers in the list. If the target is not found, returns None.
    '''

    # Recursive function to find first house with target postal code.
    def find_first_house(nums, target, left_bound, right_bound):
        '''
        Recursively finds the street number of the first house with the target in a sorted list.

        Parameters
        ----------
        nums : list of int
            A sorted list of integers representing postal codes of each house.
        target : int
            The postal code to find in the list.
        left_bound : int
            The left boundary of the search space.
        right_bound : int
            The right boundary of the search space.

        Returns
        ----------
        int or None
            The street number of the first house with the target if found, otherwise None.
        '''

        # Base case: Returns None if the search space is empty.
        if left_bound > right_bound:
            return None

        # Calculates two middle house street numbers.
        middle_house_1 = left_bound + (right_bound - left_bound) // 3
        middle_house_2 = right_bound - (right_bound - left_bound) // 3

        # Checks if any of middle houses have target, and continue searching left.
        if nums[middle_house_1] == target:
            first_house = find_first_house(nums, target, left_bound, middle_house_1 - 1)
            if first_house != None:
                return first_house
            else:
                return middle_house_1
        elif nums[middle_house_2] == target:
            first_house = find_first_house(nums, target, left_bound, middle_house_2 - 1)
            if first_house != None:
                return first_house
            else:
                return middle_house_2
        # If target is smaller than first middle house code, searches in left side.
        elif nums[middle_house_1] > target:
            return find_first_house(nums, target, left_bound, middle_house_1 - 1)
        # If target is larger than second middle house code, searches in right side.
        elif nums[middle_house_2] < target:
            return find_first_house(nums, target, middle_house_2 + 1, right_bound)
        # Searches in middle section between two middle houses.
        else:
            return find_first_house(nums, target, middle_house_1 + 1, middle_house_2 - 1)

    # Recursive function to find last house with target postal code.
    def find_last_house(nums, target, left_bound, right_bound):
        '''
        Recursively finds the street number of the last house with the target in a sorted list.

        Parameters
        ----------
        nums : list of int
            A sorted list of integers representing postal codes of each house.
        target : int
            The postal code to find in the list.
        left_bound : int
            The left boundary of the search space.
        right_bound : int
            The right boundary of the search space.

        Returns
        ----------
        int or None
            The street number of the last house with the target if found, otherwise None.
        '''

        # Base case: Returns None if the search space is empty.
        if left_bound > right_bound:
            return None

        # Calculates two middle house street numbers.
        middle_house_1 = left_bound + (right_bound - left_bound) // 3
        middle_house_2 = right_bound - (right_bound - left_bound) // 3

        # Checks if either of middle houses have target, and continue searching right.
        if nums[middle_house_1] == target:
            last_house = find_last_house(nums, target, middle_house_1 + 1, right_bound)
            if last_house != None:
                return last_house
            else:
                return middle_house_1
        elif nums[middle_house_2] == target:
            last_house = find_last_house(nums, target, middle_house_2 + 1, right_bound)
            if last_house != None:
                return last_house
            else:
                return middle_house_2
         # If target is smaller than first middle house code, searches in left side.
        elif nums[middle_house_1] > target:
            return find_last_house(nums, target, left_bound, middle_house_1 - 1)
        # If target is larger than second middle house code, searches in right side.
        elif nums[middle_house_2] < target:
            return find_last_house(nums, target, middle_house_2 + 1, right_bound)
        # Searches in middle section between two middle houses.
        else:
            return find_last_house(nums, target, middle_house_1 + 1, middle_house_2 - 1)

    # Finds first house with target postal code.
    first_house = find_first_house(nums, target, 0, len(nums) - 1)

    # Finds last house with target postal code and calculates total number of houses with target. 
    last_house = find_last_house(nums, target, 0, len(nums) - 1)

    # Return None, if no house with target is found. 
    if first_house == None or last_house == None:
        return None
        
    total_houses = last_house - first_house + 1

    # Answer.
    return [first_house, last_house, total_houses]
```

## Time Complexity Analysis

```python
# Import neccessarry libraries.
import time
import numpy as np
import matplotlib.pyplot as plt
```

### Average Runtime

#### Brute Force

##### Brute Force Average Runtime Plot

```python
# Create input array sizes from 0 to 5000.
array_size = [i for i in range(5001)]

# Create storage list for average runtimes.
average_run_times = []

# Run the experiment.
for n in array_size:
    # Randomly generate random input of size n and targer postal code. 
    nums = np.sort(np.random.randint(0, 1001, n))
    target = np.random.choice(1001)

    # Create storage list for algorithm runtimes of each input size.
    run_times = []

    # Calculate runtimes of each input across 100 trials; add results to run_times.
    for _ in range(100):
        start_time = time.time()
        bruteforce_search_extension(nums, target)
        end_time = time.time()
        run_times.append(end_time - start_time)

    # Calcuate average runtimes for each input size; add results to average_run_times.
    average_run_times.append(np.mean(run_times))

# Plot and show results.
plt.figure(figsize=(12, 6))
plt.plot(array_size, average_run_times, marker='o', linestyle='', color='b', 
         markersize=3, label="Average Runtime (100 Trials)", alpha=0.7, linewidth=2)
plt.xlabel("Input Size (n)", fontsize=14)
plt.ylabel("Average Runtime (seconds)", fontsize=14)
plt.title("Average Runtime of Brute Force Algorithm vs. Input Size", fontsize=16)
plt.grid(True, which='both', linestyle='--', linewidth=0.5)
plt.xticks(fontsize=12)
plt.yticks(fontsize=12)
plt.xlim(0, max(array_size))
plt.ylim(0, max(average_run_times) * 1.1)
plt.legend()
plt.tight_layout()
plt.show()
```

##### Brute Force Average Runtime Ratios

```python
# Indices where input size doubles
indices_to_choose = [78, 156, 312, 625, 1250, 2500, 5000]

# List to store calculated ratios.
ratios = []

# Calculate ratio between runtimes at input sizes that differ two times in size.
for i in range(len(indices_to_choose) - 1):
    ratio = average_run_times[indices_to_choose[i+1]] / average_run_times[indices_to_choose[i]]
    ratios.append(ratio)

print("Brute Force Average Runtime Ratios:")

# Print values and ratios for specified indices.
for i in range(1, len(indices_to_choose)):
    print(f"Input Size: {array_size[indices_to_choose[i]]}, "
          f"Average Runtime: {average_run_times[indices_to_choose[i]]:.8f}, "
          f"Ratio: {ratios[i-1]:.8f}")
```

```python
def bruteforce_search_extension_operation_counter(nums, target):
    ''' 
    In an ordered list of houses, finds the street numbers of the first and last house with the target postal code
    and counts how many times the target postal code appears in the list using a brute-force implementation.
    Additionally, counts the number of computational operations of the algorithm based on the input size.  

    Parameters
    ----------
    nums : list of int
        A sorted list of integers representing postal codes of each house.
    target : int
        Target postal code to find in the list.

    Returns
    ----------
    list or None
        A list consisting of the street number of the first and last house with the target postal code and 
        the number of times the target postal code appers in the list. If the target is not found, returns None.
    operations : int
        The number of operations per execution.
    '''
    operations = 0  # Operations counter.
    
    # Notes street number of first and last houses with target.
    first_house = None
    last_house = None
    operations += 2 # Operations: variable initialization.

    # Iterates through list of houses and note street numbers of first and last houses with target if found.
    for street_number, postal_code in enumerate(nums):
        operations += 1 # Operation: check if statement.
        if postal_code == target:
            operations += 1  # Operation: check if statement.
            if first_house is None: # Check whether street number of first house was noted yet; if not, record street number.
                first_house = street_number
                operations += 1 # Operation: assignment.
            last_house = street_number # Update street number of last house with target.
            operations += 1 # Operation: assignment.

    # Calculates total number of houses with target if target is found; return answers.
    if first_house == None:
        operations += 2 # Operations: check if statement and return result.
        return None, operations
    else:
        operations += 3 # Operations: enter else branch, calculate total_houses, and return result.
        total_houses = last_house - first_house + 1
        return [first_house, last_house, total_houses], operations

# Create storage for number of operations for different input sizes.
number_operations = []

# Calculate number of operations for each input sizes. 
for n in array_size:
    nums = np.sort(np.random.randint(0, 1001, n)) # Possible values in list: 0 to 1000.
    target = np.random.choice(1001) # Possible target values: 0 to 1000.
    _, operations = bruteforce_search_extension_operation_counter(nums, target)
    number_operations.append(operations)

# Plot and show results. 
plt.figure(figsize=(12, 6))
plt.plot(array_size, number_operations, marker='o', linestyle='', color='b', 
         markersize=3, label="Operations Count", alpha=0.7, linewidth=2)
plt.xlabel("Input Size (n)", fontsize=14)
plt.ylabel("Number of Operations (x)", fontsize=14)
plt.title("Brute Force Algorithm Number of Operations vs. Input Size", fontsize=16)
plt.grid(True, which='both', linestyle='--', linewidth=0.5)
plt.xticks(fontsize=12)
plt.yticks(fontsize=12)
plt.xlim(0, max(array_size))
plt.ylim(0, max(number_operations) * 1.1)
plt.legend()
plt.tight_layout()
plt.show()
```

#### Iterative Binary Search

##### Iterative Binary Search Operations Plot

```python
def iterative_binary_search_extension_operation_counter(nums, target):
    ''' 
    In an ordered list of houses, finds the street numbers of the first and last house with the target postal code
    and counts how many times the target postal code appears in the list using an iterative binary search implementation.
    Additionally, counts the number of computational operations of the algorithm based on the input size. 

    Parameters
    ----------
    nums : list of int
        A sorted list of integers representing postal codes of each house.
    target : int
        Target postal code to find in the list.

    Returns
    ----------
    list or None
        A list consisting of the street number of the first and last house with the target postal code and 
        the number of times the target postal code appers in the list. If the target is not found, returns None.
    operations : int
        The number of operations per execution.
    '''
    # Operations counter.
    operations = 0
    
    # Initializes bounds for search space.
    left_bound = 0
    right_bound = len(nums) - 1
    operations += 2 # Operations: variable initialization.

    # Notes street number of first and last houses with target.
    first_house = None
    last_house = None
    operations += 2 # Operations: variable initialization.

    operations += 1 # Operation: while statement check.
    # Finds street number of first house with target code.
    while left_bound <= right_bound:
        middle_house = (left_bound + right_bound) // 2
        operations += 1 # Operation: middle house calculation.
        if nums[middle_house] == target:
            operations += 1 # Operation: check if statement.
            first_house = middle_house
            right_bound = middle_house - 1
            operations += 2 # Operations: assignment.
        elif nums[middle_house] < target:
            operations += 1 # Operation: check elif statement.
            left_bound = middle_house + 1 
            operations += 1 # Operation: assignment.
        else:
            operations += 1 # Operation: enter else branch.
            right_bound = middle_house - 1
            operations += 1 # Operation: assignment.
    
    left_bound = 0
    right_bound = len(nums) - 1
    operations += 2 # Operation: variable reinitialization.

    operations += 1 # Operation: while statement check.
    # Finds street number of last house with target code.
    while left_bound <= right_bound:
        middle_house = (left_bound + right_bound) // 2
        operations += 1 # Operation: middle house calculation.
        if nums[middle_house] == target:
            operations += 1 # Operation: check if statement.
            last_house = middle_house
            left_bound = middle_house + 1
            operations += 2 # Operations: assignment.
        elif nums[middle_house] < target:
            operations += 1 # Operation: check elif statement. 
            left_bound = middle_house + 1
            operations += 1 # Operation: assignment.
        else:
            operations += 1 # Operation: enter else branch.
            right_bound = middle_house - 1
            operations += 1 # Operation: assignment.

    operations += 1 # Operation: check if statement
    # Returns None if target is not in list of houses.
    if first_house == None:
        operations += 1 # Operation: return result.
        return None, operations

    # Calculates number of houses with target code in list.
    total_houses = last_house - first_house + 1
    operations += 2 # Operations: total house calculation and return results.

    return [first_house, last_house, total_houses], operations

# Create storage for number of operations for different input sizes.
number_operations = []

# Calculate number of operations for each input sizes. 
for n in array_size:
    nums = np.sort(np.random.randint(0, 1001, n)) # Possible values in list: 0 to 1000.
    target = np.random.choice(1001) # Possible target values: 0 to 1000.
    _, operations = iterative_binary_search_extension_operation_counter(nums, target)
    number_operations.append(operations)

# Plot and show results. 
plt.figure(figsize=(12, 6))
plt.plot(array_size, number_operations, marker='o', linestyle='', color='b', 
         markersize=3, label="Operations Count", alpha=0.7, linewidth=2)
plt.xlabel("Input Size (n)", fontsize=14)
plt.ylabel("Number of Operations (x)", fontsize=14)
plt.title("Iterative Binary Search Algorithm Number of Operations vs. Input Size", 
          fontsize=16)
plt.grid(True, which='both', linestyle='--', linewidth=0.5)
plt.xticks(fontsize=12)
plt.yticks(fontsize=12)
plt.xlim(0, max(array_size))
plt.ylim(0, max(number_operations) * 1.1)
plt.legend()
plt.tight_layout()
plt.show()
```

#### Recursive Binary Search

##### Recursive Binary Search Operations Plot

```python
def recursive_binary_search_extension_operation_counter(nums, target):
    ''' 
    In an ordered list of houses, finds the street numbers of the first and last house with the target postal code
    and counts how many times the target postal code appears in the list using a recursive binary search implementation.
    Additionally, counts the number of computational operations of the algorithm based on the input size.

    Parameters
    ----------
    nums : list of int
        A sorted list of integers representing postal codes of each house.
    target : int
        Target postal code to find in the list.

    Returns
    ----------
    list or None
        A list consisting of the street number of the first and last house with the target postal code and 
        the number of times the target postal code appers in the list. If the target is not found, returns None.
    operations_all : int
        The number of all operations per execution of the algorithm.
    '''

    # Recursive function to find first house with target postal code.
    def find_first_house_operation_counter(nums, target, left_bound, right_bound):
        '''
        Recursively finds the street number of the first house with the target in a sorted list.
        Additionally, counts the number of computational operations of the algorithm based on the input size.

        Parameters
        ----------
        nums : list of int
            A sorted list of integers representing postal codes of each house.
        target : int
            The postal code to find in the list.
        left_bound : int
            The left boundary of the search space.
        right_bound : int
            The right boundary of the search space.

        Returns
        ----------
        int or None
            The street number of the first house with the target if found, otherwise None.
        operations_first : int
            The number of operations per execution when finding the first house.
        '''
        # Operations counter.
        operations_first = 0

        operations_first += 1 # Operation: check if statement.
        # Base case: Returns None if gone beyond bounds.
        if left_bound > right_bound:
            operations_first += 1 # Operation: return result.
            return None, operations_first

        operations_first += 1 # Operation: middle house calculation.
        # Calculates middle house street number.
        middle_house = (left_bound + right_bound) // 2

        if nums[middle_house] == target:
            operations_first += 1 # Operation: check if statement.
            # Searches left to find street number of first house with target.
            first_house, left_operations = find_first_house_operation_counter(nums, 
                                                target, left_bound, middle_house - 1)
            operations_first += left_operations # Add operations from left recursion.

            # If another house with target is found on the left, returns the street number of that house. 
            # Otherwise, returns middle house street number.
            if first_house != None:
                operations_first += 2 # Operation: check if statement and return result.
                return first_house, operations_first
            else:
                operations_first += 2 # Operations: enter else branch and return result.
                return middle_house, operations_first
                
        elif nums[middle_house] > target:
            operations_first += 2 # Operation: check elif statement and call recursive function.
            first_house, left_operations = find_first_house_operation_counter(nums,
                                                target, left_bound, middle_house - 1)
            operations_first += left_operations # Add operations from left recursion.
            return first_house, operations_first
        else:
            operations_first += 2 # Operation: enter else branch and call recursive function.
            first_house, right_operations = find_first_house_operation_counter(nums, 
                                                target, middle_house + 1, right_bound)
            operations_first += right_operations # Add operations from right recursion.
            return first_house, operations_first

    # Recursive function to find last house with target postal code.
    def find_last_house_operation_counter(nums, target, left_bound, right_bound):
        '''
        Recursively finds the street number of the last house with the target in a sorted list.
        Additionally, counts the number of computational operations of the algorithm based on the input size.

        Parameters
        ----------
        nums : list of int
            A sorted list of integers representing postal codes of each house.
        target : int
            The postal code to find in the list.
        left_bound : int
            The left boundary of the search space.
        right_bound : int
            The right boundary of the search space.

        Returns
        ----------
        int or None
            The street number of the last house with the target if found, otherwise None.
        operations_last : int
            The number of operations per execution when finding the last house.
        '''

        # Operations counter.
        operations_last = 0

        operations_last += 1 # Operation: check if statement.
        # Base case: Returns None if gone beyond bounds.
        if left_bound > right_bound:
            operations_last += 1 # Operation: return result.
            return None, operations_last

        operations_last += 1 # Operation: middle house calculation.
        # Calculates middle house street number.
        middle_house = (left_bound + right_bound) // 2

        if nums[middle_house] == target:
            operations_last += 2 # Operation: check if statement and call recursive function.
            # Searches right to find street number of first house with target.
            last_house, right_operations = find_last_house_operation_counter(nums, 
                                                target, middle_house + 1, right_bound)
            operations_last += right_operations # Add operations from right recursion.

            # If another house with target is found on the right, return the street number of that house. 
            # Otherwise, return middle house street number.
            if last_house != None:
                operations_last += 2 # Operation: check if statement and return result.
                return last_house, operations_last
            else:
                operations_last += 2 # Operation: enter else branch and return result.
                return middle_house, operations_last 
        elif nums[middle_house] > target:
            operations_last += 2 # Operation: check elif statement and call recursive function.
            last_house, left_operations = find_last_house_operation_counter(nums, 
                                                target, middle_house + 1, right_bound)
            operations_last += left_operations # Add operations from left recursion.
            return last_house, operations_last
        else:
            operations_last += 2 # Operation: enter else branch and call recursive function.
            last_house, right_operations = find_last_house_operation_counter(nums, 
                                                target, middle_house + 1, right_bound)
            operations_last += right_operations # Add operations from right recursion.
            return last_house, operations_last
    
    # Operations counter.
    operations_all = 0

    operations_all += 1 # Operation: call recursive function.
    # Finds street number of first house with target code.
    first_house, first_operations = find_first_house_operation_counter(nums, target, 0, len(nums) - 1)
    operations_all += first_operations

    operations_all += 2 # Operation: call recursive function and total house calculation.
    # Finds street number of first house with target code and calculates total number of houses with street number.
    last_house, last_operations = find_last_house_operation_counter(nums, target, 0, len(nums) - 1)
    operations_all += last_operations

    operations_all += 1 # Operation: check if statement.
    # Returns None if no house with target code is found.
    if first_house == None or last_house == None:
        operations_all += 1 # Operation: return result.
        return None, operations_all
    
    total_houses = last_house - first_house + 1

    operations_all += 2 # Operations: total house calculation and return answer.
    # Answer.
    return [first_house, last_house, total_houses], operations_all

# Create storage for number of operations for different input sizes.
number_operations = []

# Calculate number of operations for each input sizes. 
for n in array_size:
    nums = np.sort(np.random.randint(0, 1001, n)) # Possible values in list: 0 to 1000.
    target = np.random.choice(1001) # Possible target values: 0 to 1000.
    _, operations = recursive_binary_search_extension_operation_counter(nums, target)
    number_operations.append(operations)

# Plot and show results. 
plt.figure(figsize=(12, 6))
plt.plot(array_size, number_operations, marker='o', linestyle='', color='b', 
         markersize=3, label="Operations Count", alpha=0.7, linewidth=2)
plt.xlabel("Input Size (n)", fontsize=14)
plt.ylabel("Number of Operations (x)", fontsize=14)
plt.title("Recursive Binary Search Algorithm Number of Operations vs. Input Size", 
          fontsize=16)
plt.grid(True, which='both', linestyle='--', linewidth=0.5)
plt.xticks(fontsize=12)
plt.yticks(fontsize=12)
plt.xlim(0, max(array_size))
plt.ylim(0, max(number_operations) * 1.1)
plt.legend()
plt.tight_layout()
plt.show()
```

```python
def iterative_ternary_search_extension_operation_counter(nums, target):
    ''' 
    In an ordered list of houses, finds the street numbers of the first and last house with the target postal code
    and counts how many times the target postal code appears in the list using an iterative ternary search implementation.
    Additionally, counts the number of computational operations of the algorithm based on the input size. 

    Parameters
    ----------
    nums : list of int
        A sorted list of integers representing postal codes of each house.
    target : int
        Target postal code to find in the list.

    Returns
    ----------
    list or None
        A list consisting of the street number of the first and last house with the target postal code and 
        the number of times the target postal code appers in the list. If the target is not found, returns None.
    operations : int
        The number of operations per execution.
    '''

    # Operations counter.
    operations = 0
    
    # Initializes bounds for search space.
    left_bound = 0
    right_bound = len(nums) - 1
    operations += 2 # Operations: variable initialization.

    # Notes street number of first and last houses with target.
    first_house = None
    last_house = None
    operations += 2 # Operations: variable initialization.

    operations += 1 # Operation: while statement check.
    # Finds street number of first house with target code.
    while left_bound <= right_bound:
        # Calculates two middle house street numbers.
        middle_house_1 = left_bound + (right_bound - left_bound) // 3
        middle_house_2 = right_bound - (right_bound - left_bound) // 3
        operations += 2 # Operations: two middle house calculation.

        if nums[middle_house_1] == target:
            operations += 1 # Operation: check if statement.
            first_house = middle_house_1
            right_bound = middle_house_1 - 1
            operations += 2 # Operations: variable assignment.
        elif nums[middle_house_2] == target:
            operations += 1 # Operation: check elif statement.
            first_house = middle_house_2
            right_bound = middle_house_2 - 1
            operations += 2 # Operations: variable assignment.
        elif target < nums[middle_house_1]:
            operations += 1 # Operation: check elif statement.
            right_bound = middle_house_1 - 1
            operations += 1 # Operation: variable assignment.
        elif target > nums[middle_house_2]:
            operations += 1 # Operation: check elif statement.
            left_bound = middle_house_2 + 1
            operations += 1 # Operation: variable assignment.
        else:
            operations += 1 # Operation: enter else branch.
            left_bound = middle_house_1 + 1
            right_bound = middle_house_2 - 1
            operations += 2 # Operations: variable assignment.

    # Re-initialize search space.
    left_bound = 0
    right_bound = len(nums) - 1
    operations += 2 # Operations: variable reinitialization.

    operations += 1 # Operation: while statement check.
    # Finds street number of last house with target code.
    while left_bound <= right_bound:
        # Calculates two middle house street numbers.
        middle_house_1 = left_bound + (right_bound - left_bound) // 3
        middle_house_2 = right_bound - (right_bound - left_bound) // 3
        operations += 2 # Operations: two middle house calculation.

        if nums[middle_house_2] == target:
            operations += 1 # Operation: check if statement.
            last_house = middle_house_2
            left_bound = middle_house_2 + 1 
            operations += 2 # Operations: variable assignment.
        elif nums[middle_house_1] == target:
            operations += 1 # Operation: check elif statement.
            last_house = middle_house_1
            left_bound = middle_house_1 + 1  
            operations += 2 # Operations: variable assignment.
        elif target < nums[middle_house_1]:
            operations += 1 # Operation: check elif statement.
            right_bound = middle_house_1 - 1
            operations += 1 # Operation: variable assignment.
        elif target > nums[middle_house_2]:
            operations += 1 # Operation: check elif statement.
            left_bound = middle_house_2 + 1
            operations += 1 # Operation: variable assignment.
        else:
            operations += 1 # Operation: enter else branch.
            left_bound = middle_house_1 + 1
            right_bound = middle_house_2 - 1
            operations += 2 # Operation: variable assignment.

    operations += 1 # Operation: check if statement
    # Returns None if target is not in list of houses.
    if first_house == None:
        operations += 1 # Operation: return result.
        return None, operations

    # Calculates number of houses with target in list.
    total_houses = last_house - first_house + 1
    operations += 2 # Operations: total house calculation and return results.

    # Answer.
    return [first_house, last_house, total_houses], operations
    
# Create storage for number of operations for different input sizes.
number_operations = []

# Calculate number of operations for each input sizes. 
for n in array_size:
    nums = np.sort(np.random.randint(0, 1001, n)) # Possible values in list: 0 to 1000.
    target = np.random.choice(1001) # Possible target values: 0 to 1000.
    _, operations = iterative_ternary_search_extension_operation_counter(nums, target)
    number_operations.append(operations)

# Plot and show results. 
plt.figure(figsize=(12, 6))
plt.plot(array_size, number_operations, marker='o', linestyle='', color='b', 
         markersize=3, label="Operations Count", alpha=0.7, linewidth=2)
plt.xlabel("Input Size (n)", fontsize=14)
plt.ylabel("Number of Operations (x)", fontsize=14)
plt.title("Iterative Ternary Search Algorithm Number of Operations vs. Input Size", 
          fontsize=16)
plt.grid(True, which='both', linestyle='--', linewidth=0.5)
plt.xticks(fontsize=12)
plt.yticks(fontsize=12)
plt.xlim(0, max(array_size))
plt.ylim(0, max(number_operations) * 1.1)
plt.legend()
plt.tight_layout()
plt.show()
```

#### Recursive Ternary Search

##### Recursive Ternary Search Operations Plot

```python
def recursive_ternary_search_extension_operation_counter(nums, target):
    ''' 
    In an ordered list of houses, finds the street numbers of the first and last house with the target postal code
    and counts how many times the target postal code appears in the list using a recursive ternary search implementation.
    Additionally, counts the number of computational operations of the algorithm based on the input size.

    Parameters
    ----------
    nums : list of int
        A sorted list of integers representing postal codes of each house.
    target : int
        Target postal code to find in the list.

    Returns
    ----------
    list or None
        A list consisting of the street number of the first and last house with the target postal code and 
        the number of times the target postal code appers in the list. If the target is not found, returns None.
    operations_all : int
        The number of all operations per execution of the algorithm.
    '''

    # Recursive function to find first house with target postal code.
    def find_first_house_operation_counter(nums, target, left_bound, right_bound):
        '''
        Recursively finds the street number of the first house with the target in a sorted list.
        Additionally, counts the number of computational operations of the algorithm based on the input size.

        Parameters
        ----------
        nums : list of int
            A sorted list of integers representing postal codes of each house.
        target : int
            The postal code to find in the list.
        left_bound : int
            The left boundary of the search space.
        right_bound : int
            The right boundary of the search space.

        Returns
        ----------
        int or None
            The street number of the first house with the target if found, otherwise None.
        operations_first : int
            The number of operations per execution when finding the first house.
        '''

        # Operations counter.
        operations_first = 0

        operations_first += 1 # Operation: check if statement.
        # Base case: Returns None if the search space is empty.
        if left_bound > right_bound:
            operations_first += 1 # Operation: return result.
            return None, operations_first

        # Calculates two middle house street numbers.
        middle_house_1 = left_bound + (right_bound - left_bound) // 3
        middle_house_2 = right_bound - (right_bound - left_bound) // 3
        operations_first += 2 # Operations: calculate two middle houses.

        # Checks if any of middle houses have target, and continue searching left.
        if nums[middle_house_1] == target:
            operations_first += 2 # Operation: check if statement and call recursive function.
            first_house, left_operations = find_first_house_operation_counter(nums, 
                                                    target, left_bound, middle_house_1 - 1)
            operations_first += left_operations # Add operations from left recursion.
            if first_house != None:
                operations_first += 2 # Operations: check if statement and return result.
                return first_house, operations_first
            else:
                operations_first += 2 # Operations: enter else branch and return result.
                return middle_house_1, operations_first
        elif nums[middle_house_2] == target:
            operations_first += 2 # Operations: check elif statement and call recursive function.
            first_house, left_operations = find_first_house_operation_counter(nums, target, left_bound, middle_house_2 - 1)
            operations_first += left_operations # Add operations from left recursion. 
            if first_house != None:
                operations_first += 2 # Operations: check if statement and return result.
                return first_house, operations_first
            else:
                operations_first += 2 # Operations: enter else branch and return result.
                return middle_house_2, operations_first
        # If target is smaller than first middle house code, searches in left side.
        elif nums[middle_house_1] > target:
            operations_first += 2 # Operations: check elif statement and call recursive function.
            first_house, left_operations = find_first_house_operation_counter(nums, 
                                                target, left_bound, middle_house_1 - 1)
            operations_first += left_operations # Add operations from left recursion.
            operations_first += 1 # Operation: return result.
            return first_house, operations_first
        # If target is larger than second middle house code, searches in right side.
        elif nums[middle_house_2] < target:
            operations_first += 2 # Operations: check elif statement and call recursive function.
            first_house, right_operations = find_first_house_operation_counter(nums, 
                                                target, middle_house_2 + 1, right_bound)
            operations_first += right_operations # Add operations from right recursion.
            operations_first += 1 # Operation: return result.
            return first_house, operations_first
        # Searches in middle section between two middle houses.
        else:
            operations_first += 2 # Operations: enter else branch and call recursive function.
            first_house, middle_operations = find_first_house_operation_counter(nums, 
                                                target, middle_house_1 + 1, middle_house_2 - 1)
            operations_first += middle_operations # Add operations from right recursion.
            operations_first += 1 # Operation: return result.
            return first_house, operations_first

    # Recursive function to find last house with target postal code.
    def find_last_house_operation_counter(nums, target, left_bound, right_bound):
        '''
        Recursively finds the street number of the last house with the target in a sorted list.
        Additionally, counts the number of computational operations of the algorithm based on the input size.

        Parameters
        ----------
        nums : list of int
            A sorted list of integers representing postal codes of each house.
        target : int
            The postal code to find in the list.
        left_bound : int
            The left boundary of the search space.
        right_bound : int
            The right boundary of the search space.

        Returns
        ----------
        int or None
            The street number of the last house with the target if found, otherwise None.
        operations_last : int
            The number of operations per execution when finding the last house.
        '''
        # Operations counter.
        operations_last = 0

        operations_last += 1 # Operation: check if statement.
        # Base case: Returns None if the search space is empty.
        if left_bound > right_bound:
            operations_last += 1 # Operation: return result.
            return None, operations_last

        # Calculates two middle house street numbers.
        middle_house_1 = left_bound + (right_bound - left_bound) // 3
        middle_house_2 = right_bound - (right_bound - left_bound) // 3
        operations_last += 2 # Operations: calculate two middle houses.

        # Checks if either of middle houses have target, and continue searching right.
        if nums[middle_house_1] == target:
            operations_last += 2 # Operations: check if statement and call recursive function.
            last_house, right_operations = find_last_house_operation_counter(nums,
                                            target, middle_house_1 + 1, right_bound)
            operations_last += right_operations # Add operations from right recursion.
            if last_house != None:
                operations_last += 2 # Operations: check if statement and return result.
                return last_house, operations_last
            else:
                operations_last += 2 # Operations: enter else branch and return result.
                return middle_house_1, operations_last
        elif nums[middle_house_2] == target:
            operations_last += 2 # Operations: check if statement and call recursive function.
            last_house, right_operations = find_last_house_operation_counter(nums, 
                                            target, middle_house_2 + 1, right_bound)
            operations_last += right_operations
            if last_house != None:
                operations_last += 2 # Operations: check if statement and return result.
                return last_house, operations_last
            else:
                operations_last += 2 # Operations: check if statement and return result.
                return middle_house_2, operations_last
         # If target is smaller than first middle house code, searches in left side.
        elif nums[middle_house_1] > target:
            operations_last += 2 # Operations: check if statement and call recursive function.
            last_house, left_operations = find_last_house_operation_counter(nums,
                                            target, left_bound, middle_house_1 - 1)
            operations_last += left_operations # Add operations from left recursion. 
            operations_last += 1 # Operation: return result.
            return last_house, operations_last
        # If target is larger than second middle house code, searches in right side.
        elif nums[middle_house_2] < target:
            operations_last += 2 # Operations: check if statement and call recursive function.
            last_house, right_operations = find_last_house_operation_counter(nums, 
                                            target, middle_house_2 + 1, right_bound)
            operations_last += right_operations # Add operations from right recursion.
            operations_last += 1 # Operation: return result.
            return last_house, operations_last
        # Searches in middle section between two middle houses.
        else:
            operations_last += 2 # Operations: enter else branch and call recursive function.
            last_house, middle_operations = find_last_house_operation_counter(nums, 
                                        target, middle_house_1 + 1, middle_house_2 - 1)
            operations_last += middle_operations # Add operations from middle recursion.
            operations_last += 1 # Operation: return result.
            return last_house, operations_last

    # Operations counter.
    operations_all = 0

    operations_all += 1 # Operation: call recursive function.
    # Finds first house with target postal code.
    first_house, operations_first = find_first_house_operation_counter(nums, target, 0, len(nums) - 1)
    operations_all += operations_first # Add operations from first house function.

    operations_all += 1 # Operation: call recursive function.
    # Finds last house with target postal code and calculates total number of houses with target. 
    last_house, operations_last = find_last_house_operation_counter(nums, target, 0, len(nums) - 1)
    operations_all += operations_last # Add operations from last house function.

    operations_all += 1 # Operation: check if statement.
    # Return None, if no house with target is found. 
    if first_house == None or last_house == None:
        operations_all += 1 # Operation: return result.
        return None, operations_all
        
    total_houses = last_house - first_house + 1

    operations_all += 2 # Operations: total house calculation and return result.
    # Answer.
    return [first_house, last_house, total_houses], operations_all

# Create storage for number of operations for different input sizes.
number_operations = []

# Calculate number of operations for each input sizes. 
for n in array_size:
    nums = np.sort(np.random.randint(0, 1001, n)) # Possible values in list: 0 to 1000.
    target = np.random.choice(1001) # Possible target values: 0 to 1000.
    _, operations = recursive_ternary_search_extension_operation_counter(nums, target)
    number_operations.append(operations)

# Plot and show results. 
plt.figure(figsize=(12, 6))
plt.plot(array_size, number_operations, marker='o', linestyle='', color='b', 
         markersize=3, label="Operations Count", alpha=0.7, linewidth=2)
plt.xlabel("Input Size (n)", fontsize=14)
plt.ylabel("Number of Operations (x)", fontsize=14)
plt.title("Recursive Ternary Search Algorithm Number of Operations vs. Input Size", 
          fontsize=16)
plt.grid(True, which='both', linestyle='--', linewidth=0.5)
plt.xticks(fontsize=12)
plt.yticks(fontsize=12)
plt.xlim(0, max(array_size))
plt.ylim(0, max(number_operations) * 1.1)
plt.legend()
plt.tight_layout()
plt.show()
```
